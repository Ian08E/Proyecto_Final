```python
import math

hombres = math.comb(8, 3)
mujeres = math.comb(4, 2)

casos_favorables = hombres * mujeres
total_comites = math.comb(12, 5)

probabilidad = casos_favorables / total_comites

print("Casos favorables:", casos_favorables)
print("Total comités:", total_comites)
print("Probabilidad:", round(probabilidad, 4))
```

```python
from scipy.stats import poisson

lambda_3m = 2 * 3

prob_mas4 = 1 - poisson.cdf(4, lambda_3m)
print(round(prob_mas4, 4))

prob_exact4 = poisson.pmf(4, lambda_3m)
print(round(prob_exact4, 4))
```

```python
# Datos del problema
P_A, P_B, P_C = 0.40, 0.30, 0.30
P_def_A, P_def_B, P_def_C = 0.02, 0.03, 0.05

# Probabilidad total de defecto
P_def = P_A * P_def_A + P_B * P_def_B + P_C * P_def_C

# Bayes: probabilidad de que un producto defectuoso venga de C
prob_C_dado_def = (P_C * P_def_C) / P_def

print("Probabilidad total de defecto:", round(P_def, 4))
print("P(C | defecto):", round(prob_C_dado_def, 4))
```

```python
from scipy.stats import norm

media = 75
desv = 10

# Percentil 90
percentil_90 = norm.ppf(0.90, loc=media, scale=desv)

# Z del percentil 90
percentil_90_z = norm.ppf(0.90)

# Nuevo escenario: top 5%
# Acumulado izquierdo = 95%
percentil_95 = norm.ppf(0.95, loc=media, scale=desv)

print("Percentil 90 (nota mínima):", round(percentil_90, 4))
print("Z para percentil 90:", round(percentil_90_z, 2))
print("Percentil 95 (nueva nota mínima):", round(percentil_95, 4))
```
